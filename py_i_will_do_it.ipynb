```python
import threading
import time

def download_file(file_name):
  print(f"{file_name} ka download start hua...")
  time.sleep(2)       # 2 second wait (jaisa download ho raha ho)
  print(f"{file_name} ska download complete")

  # 3 threads banaye
  t1=threading.Thread(target=download_file,args=("File-1",))
  t1=threading.Thread(target=download_file,args=("File-2",))
  t1=threading.Thread(target=download_file,args=("File-3",))

  # start kar diya
  t1.start()
  t2.start()
  t3.start()

  # wait till all finish
  t1.join()
  t2.join()
  t3.join()

  print("Sab download complete ho gaye")
```

```python
import multiprocessing
import time

def calculate_square(n):
  print(f"square calculated kar rahe hain: {n}")
  time.sleep(1)         # calculation ka delay samjho
  print(f"{n}^2={n*n}")

if __name__ =="__main__":
  numbers=[2,3,4]

# har number ka square nikalne ke liye alag process
processes=[]
for num in numbers:
  p=multiprocessing.Process(target=calculate_square,args=(num,))
  processes.append(p)
  p.start()
```

```python
import pandas as pd
import numpy as np

from sklearn.utils import resample

# step 1:create imbalance dataset
data={
    'Feature_1':np.random.randn(1000),
    'Feature_2':np.random.randn(1000),
    'target':[0]*950+[1]*50
}

df=pd.DataFrame(data)

print("Original class distribution:")
print(df['target'].value_counts())

# step 2: split majority and minority classess
class_0=df[df['target']==0]     # Majority
class_1=df[df['target']==1]     # Minority

# Step 3: Upsample minority class
class_1_upsampled=resample(
    class_1,
    replace=True,
    n_samples=len(class_0),
    random_state=42
)

# Step 4: Combine majority  and upsampled minority
df_upsampled=pd.concat([class_0,class_1_upsampled])

print("\nAfter upsampling class distribution:")
print(df_upsampled['target'].value_counts())
```

```python
import pandas as pd
import numpy as np
from sklearn.utils import resample

# Step 1: create imbalance dataset
data={
    'Feature_1':np.random.randn(1000),
    'Feature_2':np.random.randn(1000),
    'target':[0]*950+[1]*50
}
df=pd.DataFrame(data)

print("Original class distribution:")
print(df['target'].value_counts())

# Step 2: splits majority and minority classess
class_0=df[df['target']==0]
class_1=df[df['target']==1]

# Step 3: Downsample majority class
class_0_downsampled=resample(
    class_0,
    replace=False,
    n_samples=len(class_1),
    random_state=42

)

# Step 4: Combine minority + downsampled majority
df_downsampled=pd.concat([class_0_downsampled,class_1])


print("\nAfter downsampling class distribution:")
print(df_downsampled['target'].value_counts())
```
